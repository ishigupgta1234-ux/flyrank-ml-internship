Precision@k, not overall accuracy. The content team can only review a
fixed number of pages a week, so what matters is whether the pages at the
top of the list are actually the right ones — not how the model does on
the thousands of pages nobody's going to look at anyway. Accuracy would
reward the model for correctly ignoring pages that were never going to be
reviewed regardless.

## 5. Why ML beats a fixed rule here

*What makes the pattern too messy for an if-statement?*

I already have evidence for this from Week 2 — a plain rule (stale ×
visible) got beaten by a simple decision tree on both Precision@20
(0.550 → 0.700) and Precision@50 (0.600 → 0.720). A fixed rule can
multiply two or three signals together, but it can't learn where the
real cutoffs are or how signals trade off against each other — like

a model picks up and a hand-written rule can't.
