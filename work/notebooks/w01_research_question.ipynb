## 2. The question: decision, action, cost of a wrong call

*What decision does your work improve? Who acts on it? What does a wrong recommendation cost?*

Decision: which pages a content reviewer should look at first when they
only have time for a handful each week.

Unit of analysis: one page.

Output: not a single score — a short ranked list with a reason attached to
each page (something like "still ranking okay but CTR has been sliding" or
"hasn't been touched in months and impressions are climbing").

Action: whoever owns content picks pages off the top of that list and
decides what to actually do — leave it alone, rewrite it, merge it with
something else, or just keep an eye on it.

Cost of getting it wrong: two different ways. If a fine page gets flagged,
someone spends an afternoon on something that didn't need it. If a page
that's actually declining gets missed, it keeps quietly losing visibility
while attention goes elsewhere — that one's worse because it compounds.

Why ML helps at all: nobody's going to manually cross-check position, CTR,
and update-recency for thousands of pages every week. A model can apply the
same rule to all of them consistently — that's really the boring, repeatable
part of the job a person shouldn't have to do by hand.

## 3. Quick look at the data (2-3 real numbers)

*Load the starter CSV below and show 2-3 real numbers that make your lane look worth the next 7 weeks.*

```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.
import pandas as pd

url = "https://raw.githubusercontent.com/ishigupgta1234-ux/flyrank-ml-internship/refs/heads/main/data/raw/content_refresh_anonymized.csv"
df = pd.read_csv(url)

print(df.shape)
print(df.columns.tolist())

# 1. How many pages are both stale and still getting real traffic
stale = df["days_since_last_update"] >= 180
visible = df["impressions_90d"] >= 500
print(f"Stale + still visible pages: {(stale & visible).sum()} out of {len(df)} "
      f"({(stale & visible).mean()*100:.1f}%)")

# 2. Which content_type underperforms most consistently on CTR
ctr_by_type_position = df.groupby(["content_type", "position_tier"])["ctr"].mean().reset_index()
worst_per_tier = ctr_by_type_position.loc[ctr_by_type_position.groupby("position_tier")["ctr"].idxmin()]
print(worst_per_tier)
```

I loaded the starter CSV directly instead of trusting my memory of the
numbers from earlier notebooks. The stale-and-visible count above is
exactly the population this lane is meant to act on. I also re-ran the
CTR-by-content-type check from Week 1 — comparison articles come out worst
in almost every position tier, a second independent signal pointing the
same direction. And from Week 2, a plain hand-rule already got beaten by a
simple decision tree on Precision@20 (0.550 → 0.700) and Precision@50
(0.600 → 0.720) — so there's real headroom for a model to add value here,
not just match what a one-line rule could already do.
